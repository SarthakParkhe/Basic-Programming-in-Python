```python
# Print sum of all digits oa a given number

num = int(input("Enter the Number"))

num =  abs(num)                        # abs means absolute value (number without negative sign)--> Absolute value of num = abs(num)
sum = 0                                # we create varaible called sum and initially given with a value 0, because we are adding digitd to it 
                                        # for 1245 start -> sum = 0
                                                        # add 1 = 1
                                                        # add 2 = 1 = 2
                                                        ....
                                                        # add 5

while num > 0:
    digit = num % 10            # % gives us the remainder to know last digit of a number here 1245/10 = 114 and remainder = 5 
    sum = sum + digit
    num = num // 10
print("Sum of digits =", sum)
```

```python
# Find factorial of given number
num = int(input("Enter the number:"))

factorial = 1  # starting with 1 ensures that multiplication works correctly


for i in range(1, num + 1):
    factorial = factorial * i

print("Factorial =", factorial)
```

```python
# Check whether a number is prime or not

num = int(input("Enter the Number:"))
if num < 2:
     print("Not a Prime Number")
else:
    prime = True     # it is assuming a number prime until proven in next steps

    for i in range(2, num):   # this loops all the integers from 2 to {num - 1} 
        if num % i == 0:    # if remainder is 0 and divisor is available it is already a composite number
            prime = False
            break 

    if prime:
        print("Prime Number")
    else:
        print("Not a Prime Number")
```

```python
# print prime numbers between 10 and 99 
print("Prime Numbers beween 10 and 99  are:")

for num in range(10, 100):
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num, end=" ")
```

```python
 # Reverse a given number
num = int(input("Enter the Number:"))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed number =", reverse)

```

```python
# Determine whether a year is leap year
year = int(input("Enter the Year"))

if year % 400 == 0:
    print("Leap Year")
elif year % 100:
    print("Not a Leap Year")
elif year % 4 == 0:
    print("Leap Year")
else: 
    print("Not a Leap Year")
```

```python
# Display Grade using if elif else
marks = int(input("Enter marks in mathmatics"))

if marks >= 90 and marks <= 100:
    grade = "O"
elif marks >= 80 and marks <= 90:
    grade = "A+"
elif marks >= 70 and marks <= 80:
    grade = "A"
elif marks >= 60 and marks <= 70:
    grade = "B"
elif marks >= 50 and marks <= 60:
    grade = "C"
elif marks >= 40 and marks <= 50:
    grade = "P"
elif marks < 40:
    grade = "F"
else:
    grade = "Invalid Marks"

print("Grade is ", grade)
```
